```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"🚀 Loading model to device: {device}")
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/umt5-base"
cache_dir = "./models"
model_path = '../models/umt5-base'

# Download and cache model + tokenizer files locally
# tokenizer = AutoTokenizer.from_pretrained(model_name, cache_dir=cache_dir)
# model = AutoModelForSeq2SeqLM.from_pretrained(model_name, cache_dir=cache_dir)

# To load the model and tokenizer:
tokenizer = AutoTokenizer.from_pretrained(model_path, local_files_only= True)
model = AutoModelForSeq2SeqLM.from_pretrained(model_path,local_files_only = True)
```

```python
model= model.to(device)
```

```python
text = "A <extra_id_0> walks into a bar and orders a <extra_id_1> with <extra_id_2> pinch of <extra_id_3>"

inputs = tokenizer(text, return_tensors="pt")

inputs= inputs.to(device)

outputs = model.generate( **inputs )
result = tokenizer.batch_decode(outputs)

print(result)
['<pad><extra_id_0>nyone who<extra_id_1> drink<extra_id_2> a<extra_id_3> alcohol<extra_id_4> A<extra_id_5> A. This<extra_id_6> I<extra_id_7><extra_id_52><extra_id_53></s>']
```
